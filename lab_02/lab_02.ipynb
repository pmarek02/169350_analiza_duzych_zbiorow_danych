````python
# obserwuj komunikaty, aby zainstalować ewentualnie brakujące komponenty
# w komórce wynikowej notebooka pojawi się link do dashboardu, w którym można obserwować aktualne
# zadania i obciążenie klastra jak, i zadania wykonane
# UWAGA! Po doinstalowaniu biblioteki bokeh może być konieczne zrestartowanie jądra Pythona i całego notebooka
# aby dashboard poprawnie działał. Każde kolejne uruchomienie poniższego kodu bez restartu jądra, utworzy nową
# instancję klienta

from dask.distributed import Client

# bez określania limitu pamięci na worker
# client = Client(n_workers=4)

# z określeniem limitu pamięci na worker
# 4 workery * 1GB = 4GB RAM - dostosuj do możliwości Twojego sprzętu
client = Client(n_workers=4, memory_limit="1GiB")

client
````

Możemy również określić limit pamięci na worker, co jest dobrym pomysłem, gdyż pozostawienie tego parametru z wartością domyślną rozdzieli pamięć po równo na każdy z workerów.
````python
import numpy as np
import dask.array as da
````

Tablica numpy oraz obliczenie średniej z tej tablicy wartości próbkowanych z rozkładu normalnego o danych parametrach.

````python
%%time
arr = np.random.normal(5, 0.2, size=(20_000, 20_000))
arr_mean = arr.mean(axis=0)
arr_mean
````

Teraz ta sama operacja z wykorzystaniem tablicy Dask.

````python
# tutaj określimy wielkość chunka ręcznie
darr = da.random.normal(5, 0.2, size=(20_000, 20_000), chunks=(2000, 2000))
darr
````

````python
darr_mean = darr.mean(axis=0)
darr_mean
````

````python
# uwaga z wywołaniem dla dużej liczby chunków tablicy dask!
darr_mean.visualize()
````

## Rozwiązania

````python
# cwiczenie
# numpy na calej tablicy ~49 GB. u mnie empty() przeszlo, od razu kasuje
# chunk z tresci (30, 4, 1000, 1000) ma ~458 MB, a ma byc 50-150
# biore (5, 4, 1000, 1000), to jest ~76 MB

import numpy as np
import dask.array as da
from datetime import datetime
from dask.distributed import Client

ksztalt = (365, 4, 3000, 3000)
print('GiB', round(365 * 4 * 3000 * 3000 * 4 / 1024 ** 3, 2))

try:
    x = np.empty(ksztalt, dtype=np.float32)
except np.core._exceptions._ArrayMemoryError as e:
    print(type(e).__name__, e)
else:
    print('empty przeszlo, GiB', round(x.nbytes / 1024 ** 3, 2))
    del x

# 4 workery po 1 GB
client = Client(n_workers=4, threads_per_worker=1, memory_limit='1GiB')
print(client.dashboard_link)

sat = da.random.uniform(0, 1, size=ksztalt, chunks=(5, 4, 1000, 1000)).astype('float32')
print(sat)

eps = 1e-6
red = sat[:, 2]
nir = sat[:, 3]
ndvi = (nir - red) / (nir + red + eps)
print(ndvi)
print('NDVI GiB', round(ndvi.nbytes / 1024 ** 3, 2))

start = datetime.now()
srednia = ndvi.mean(axis=(1, 2)).compute()
print(f'Czas: {datetime.now() - start}')
print(srednia.shape)
print(np.round(srednia[:5], 4))

start = datetime.now()
odsetek = float((ndvi < 0.2).mean().compute())
print(f'Czas: {datetime.now() - start}')
print(odsetek)

# przy 1 GB na worker pamiec zostala na kilkuset MB, nie na 49 GB
# ndvi.compute() zbieraloby ~12 GB do pythona i by nie weszlo
client.close()
````

````python
# zadanie 1
# NaN nie w pierwszych 10 wierszach
# jak pandas zapisze NaN w id, w csv jest 10248.0 i dask od razu widzi float
# liczby zostawiam jako int, puste pole tylko tam gdzie NaN

import os
import pandas as pd
import dask.dataframe as dd

df = pd.read_csv(os.path.join('..', 'lab_01', 'zamowienia.csv'), sep=';')
print(df.dtypes)
print('wierszy', len(df))

df.loc[[20, 120, 400], 'idZamowienia'] = np.nan
df.loc[[30, 250], 'Utarg'] = np.nan

zapis = df.copy()
zapis['idZamowienia'] = ['' if pd.isna(v) else str(int(v)) for v in zapis['idZamowienia']]
zapis.to_csv('zamowienia_missing.csv', sep=';', index=False)

for n in (10, 30, 200):
    ramka = dd.read_csv('zamowienia_missing.csv', sep=';', sample_rows=n)
    print('sample_rows', n)
    print(ramka.dtypes)
    try:
        cala = ramka.compute()
        print('ok', len(cala), round(float(cala['Utarg'].sum()), 2))
    except Exception as e:
        print(type(e).__name__, str(e).split('\n')[0])

ramka = dd.read_csv('zamowienia_missing.csv', sep=';', sample_rows=10, assume_missing=True)
print('assume_missing')
print(ramka.dtypes)
print('ok', round(float(ramka['Utarg'].sum().compute()), 2))
````
